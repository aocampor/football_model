# Load Results CSV
This notebook loads the `results.csv` file using pandas and displays the first few rows.

```python
import pandas as pd
import numpy as np
import lightgbm as lgbm
import matplotlib.pyplot as plt
```

```python
def read_data():
    df_ = pd.read_csv('data/results.csv')
    df_ = df_.drop(["country", "city"], axis=1)
    df_names_ = pd.read_csv('data/former_names.csv')
    df_wc_ = pd.read_csv('data/world_champions.csv')
    df_goals_ = pd.read_csv('data/goalscorers.csv')
    return df_, df_names_, df_wc_, df_goals_
```

```python
df_original, df_names_original, df_wc_original, df_goals_original = read_data()
```

```python
def calculate_time_features(df_, df_wc_):
    df_['date'] = pd.to_datetime(df_['date'])
    df_wc_['date'] = pd.to_datetime(df_wc_['date'])
    df_['year'] = df_['date'].dt.year
    df_['month'] = df_['date'].dt.month
    def get_season(month):
        if month in [12, 1, 2]:
            return 0
        elif month in [3, 4, 5]:
            return 1
        elif month in [6, 7, 8]:
            return 2
        else:
            return 3
    df_['season'] = df_['month'].apply(get_season)
    df_['day'] = df_['date'].dt.day
    df_['day_of_week'] = df_['date'].dt.dayofweek
    return df_, df_wc_
```

```python
df, df_wc = calculate_time_features(df_original, df_wc_original)
df.tail()
```

```python
def calculate_tournament_feature(df_):
    df_["tournament"] = np.where(
        df_.tournament == 'Friendly', 
        0, 
        np.where(
            df_.tournament == 'UEFA Euro', 
            1,
            np.where(
                df_.tournament == 'Copa América', 
                2,
                np.where(
                    df_.tournament == 'FIFA World Cup',
                    3, 
                    4
                )
            )
        )
    )
    return df_
```

```python
df = calculate_tournament_feature(df)
df.tail()
```

```python
def reformat_features(df_):
    df_.loc[df_.home_team == "DR Congo", "home_team"] = "Congo"
    df_.loc[df_.away_team == "DR Congo", "away_team"] = "Congo"
    df_copy = df_.copy()
    df_["team"] = df_["home_team"]
    df_["state"] = 1
    df_["opponent"] = df_["away_team"]
    df_["goals"] = df_["home_score"]
    df_["goals_allowed"] = df_["away_score"]

    df_copy["team"] = df_copy["away_team"]
    df_copy["state"] = 0
    df_copy["opponent"] = df_copy["home_team"]
    df_copy["goals"] = df_copy["away_score"]
    df_copy["goals_allowed"] = df_copy["home_score"]

    df_ = pd.concat([df_, df_copy])
    df_ = df_.drop(
        columns=[
            "home_team", "away_team", "home_score", 
            "away_score"
        ]
    )
    df_["neutral"] = df_["neutral"].astype(int) 
    return df_
```

```python
df = reformat_features(df)
df.tail()
```

```python
def get_scorers(df_):
    df_['date'] = pd.to_datetime(df_['date'])
    output = []
    for (date_, home_team_, away_team_), grouped_data in df_.groupby(
        ["date", "home_team", "away_team"]
    ):
        feats = ["team", "scorer", "date"]
        grouped_scorers = grouped_data[feats].groupby(["team", "scorer"]).count().reset_index()
        home_scorers = len(grouped_scorers[grouped_scorers.team == home_team_])
        away_scorers = len(grouped_scorers[grouped_scorers.team == away_team_])
        out_df = pd.DataFrame(
            {
                "date": [date_, date_],
                "team": [home_team_, away_team_],
                "opponent": [away_team_, home_team_],
                "scorers": [home_scorers, away_scorers],
                # "opponent_scorers": [away_scorers, home_scorers]
            }
        )
        output.append(
            out_df
        )
    return pd.concat(output)
```

```python
df_scores = get_scorers(df_goals_original)
df_scores.tail()
```

```python
def calculate_points(df_):
    df_['goal_diff'] = df_['goals'] - df_['goals_allowed']
    df_['points'] = np.where(
        df_['goal_diff'] > 0, 
        3, 
        np.where(
            df_['goal_diff'] == 0, 
            1, 
            0
        )
    )
    return df_
```

```python
df = calculate_points(df)
to_shift.append("goal_diff")
to_shift.append("points")
df.head()
```

```python
def calculate_aggreated_features(df_):
    grouped_teams = df_.groupby(["team", "state"])
    list_df_out = []
    aggregate_window = 2
    for (team, state), group in grouped_teams:
        group = group.sort_values("date", ascending=True)
        group["cum_points"] = np.cumsum(group["points"])
        group["window_cum_points"] = group["points"].rolling(window=aggregate_window).sum()
        group["window_mean_points"] = group["points"].rolling(window=aggregate_window).mean()

        group["average_goals_per_game"] = group["goals"].rolling(window=aggregate_window).mean()
        group["average_goals_allowed_per_game"] = group["goals_allowed"].rolling(window=aggregate_window).mean()

        group["scorers_mean"] = group["scorers"].rolling(window=aggregate_window).mean()
        
        group["average_diff_goals_per_game"] = group["average_goals_per_game"] - group["average_goals_allowed_per_game"]
        
        list_df_out.append(group)
    return pd.concat(list_df_out)
```

```python
df = calculate_aggreated_features(df)
aggregated_metrics = [
    "cum_points", "window_cum_points", "window_mean_points", 
    "average_goals_per_game", "average_goals_allowed_per_game",
    "scorers_mean", "average_diff_goals_per_game"
]
to_shift = to_shift + aggregated_metrics
df.head()
```

```python
def calculate_opponent_features(df_):
    df_ = df_.sort_values("date", ascending=True)
    df_ = pd.merge(
        df_, 
        df_[
            ["team", "date", 
             "goals", "goals_allowed", "goal_diff", 
             "points", "cum_points", "window_cum_points", "window_mean_points",
             "average_goals_per_game", "average_goals_allowed_per_game",
             "average_diff_goals_per_game", 
             "scorers", "scorers_mean"
            ]
        ].rename(
            columns={
                "team": "opponent",
                "goals": "opponent_goals",
                "goals_allowed": "opponent_goals_allowed",
                "goal_diff": "opponent_goal_diff",
                "points": "opponent_points",
                "cum_points": "opponent_cum_points", 
                "window_cum_points": "opponent_window_cum_points",
                "window_mean_points": "opponent_window_mean_points",
                "average_goals_per_game": "opponent_average_goals_per_game",
                "average_goals_allowed_per_game": "opponent_average_goals_allowed_per_game",
                "average_diff_goals_per_game": "opponent_average_diff_goals_per_game",
                "scorers": "opponent_scorers",
                "scorers_mean": "opponent_scorers_mean"
            }
        ), 
        on=["opponent", "date"], 
        how="left"
    )
    df_["diff_cum_points"] = df_.cum_points - df_.opponent_cum_points
    df_["diff_window_cum_points"] = df_.window_cum_points - df_.opponent_window_cum_points
    df_["diff_window_mean_points"] = df_.window_mean_points - df_.opponent_window_mean_points
    df_["target"] = df_["goal_diff"]
    return df_
```

```python
df = calculate_opponent_features(df)
opponent_features = [
    "opponent_goals", "opponent_goals_allowed", "opponent_goal_diff",
    "opponent_points", "opponent_cum_points", "opponent_window_cum_points",
    "opponent_window_mean_points", "opponent_average_goals_per_game",
    "opponent_average_goals_allowed_per_game", 
    "opponent_average_diff_goals_per_game", "opponent_scorers",
    "opponent_scorers_mean", "diff_cum_points", "diff_window_cum_points",
    "diff_window_mean_points"
]
to_shift = to_shift + opponent_features
df.head()
```

```python
def get_wc_features(df_, df_wc_):
    df_ = pd.merge(
        df_, 
        df_wc_, 
        on=["team", "date"], 
        how="left"
    )
    df_['wc'] = np.where(df_['position'] == 1, 1, 0)
    df_['wc'] = df_.sort_values('date', ascending=True).groupby('team')['wc'].cumsum()
    return df_
```

```python
df = get_wc_features(df, df_wc_original)
categorical_features += ['wc']
df.head()
```

```python
def shift_features(df_, to_shift_):
    for feat in to_shift_:
        df_[feat] = df_[feat].shift(1)
    return df_
```

```python
df = shift_features(df, to_shift + ['wc'])
feature_names = categorical_features + to_shift
df.head()
```

```python
def train(df_, feature_names, target, categorical_cols):
    n_min = 365*10
    n_ = 180
    n_val = n_
    n_test = n_
    n_total = len(df_.date.unique())
    df_ = df_.sort_values("date", ascending=True)

    val_scores = []
    test_scores = []
    baseline_scores = []

    for i in range(n_min, n_total - n_val, n_val):
        try:
            print(
                df_.date.unique()[i], 
                df_.date.unique()[i + n_val], 
                df_.date.unique()[i + n_val + n_test]
            )
        except:
            print("Last")
        df_train = df_[
            df_.date <= df_.date.unique()[i]
        ]
        df_val = df_[
            (df_.date > df_.date.unique()[i]) & 
            (df_.date <= df_.date.unique()[i + n_val])
        ]
        try:
            df_test = df_[
                (df_.date > df_.date.unique()[i + n_val]) & 
                (df_.date <= df_.date.unique()[i + n_val + n_test])
            ]
        except:
            print("Last")
        print("before features")
        features_train = df_train[feature_names].copy()
        features_val = df_val[feature_names].copy()
        try:
            features_test = df_test[feature_names].copy()
        except:
            print("Last")
            
        for col in categorical_cols:
            features_train[col] = features_train[col].astype('category')
            features_val[col] = features_val[col].astype('category')
            try:
                features_test[col] = features_test[col].astype('category')
            except:
                print("Last")
        print("before target")
        labels_train = df_train["target"]
        labels_val = df_val["target"]
        try:
            labels_test = df_test["target"]
        except:
            print("Last")
        
        if target == "points":
            labels_train = labels_train.astype('category').copy()
            labels_val = labels_val.astype('category').copy()
            try:
                labels_test = labels_test.astype('category').copy()
            except:
                print("Last")

        X_train = features_train
        y_train = labels_train
        
        X_val = features_val
        y_val = labels_val

        try:
            X_test = features_test
            y_test = labels_test
        except:
            print("Last")
        
        if target == "points":
            model = lgbm.LGBMClassifier(
                objective='multiclass',
                num_class=3,
                metric='multi_logloss',
                is_unbalance=True,
                n_estimators=10000, 
            )
            early_stopping = lgbm.early_stopping(stopping_rounds=50)
            log_evaluation = lgbm.log_evaluation(period=10)
            model.fit(
                X_train, y_train,
                eval_set=[(X_val, y_val)],  # Crucial: Early stopping needs a validation set
                callbacks=[early_stopping, log_evaluation]
            )
        #elif target == "goal_diff":
        else:
            model = lgbm.LGBMRegressor(
                objective="regression",
                metric="rmse"
            )
            model.fit(X_train, y_train)
        y_pred = model.predict(X_val)
        print(f"Validation score: {model.score(X_val, y_val)}")
        val_scores.append(model.score(X_val, y_val))
        baseline_scores = [1 if x == 0 else 0 for x in y_val]
        try: 
            y_pred_test = model.predict(X_test)
            print(f"Test score: {model.score(X_test, y_test)}")
            test_scores.append(model.score(X_test, y_test))
        except:
            print("Last")
        plt.figure(figsize=(10, 8))
        lgbm.plot_importance(model, importance_type='gain', figsize=(10, 8))
        plt.show()
    baseline_metric = np.mean(baseline_scores)
    plt.figure(figsize=(10, 6))
    plt.hist(
        val_scores[:-1], 
        label='Validation', alpha=0.5, 
        weights=np.ones(len(val_scores[:-1])) / len(val_scores[:-1])
    )
    plt.hist(   
        test_scores, 
        label='Test', alpha=0.5, 
        weights=np.ones(len(test_scores)) / len(test_scores)
    )
    plt.legend()
    plt.show()
    plt.figure(figsize=(10, 6))
    plt.title(f"baseline score {baseline_metric}")
    plt.plot(
        val_scores[:-1], 
        label=f'Validation {np.mean(val_scores[:-1])}', alpha=0.5,
    )
    plt.plot(   
        test_scores, 
        label=f'Test {np.mean(test_scores)}', alpha=0.5,
    )
    plt.legend()
    plt.show()
    return model
```

```python
def preprocessing(df_, df_wc_, df_scorers_):
    df_, df_wc_ = calculate_time_features(df_, df_wc_)
    df_ = calculate_tournament_feature(df_)
    df_ = reformat_features(df_)
    df_scores_ = get_scorers(df_scorers_)
    df_ = pd.merge(df_, df_scores_, how="left", on=["date", "team", "opponent"])
    categorical_features = [
        "tournament", "neutral", "month",
        "season", "day", "day_of_week",
        "state"
    ]
    to_shift = [
        "goals", "goals_allowed", "scorers"
    ]
    target = "goal_diff"
    df_ = calculate_point_features(df_)
    print(df_.tail())
    df_ = get_wc_features(df_, df_wc_)
    df_ = df_.dropna()
    # df_ = df_.fillna(0)
    return df_, df_wc_
```

```python
def generate_prediction_dataset(games_, df_, df_wc_, model, features, categoricals):
    for row in games_.itertuples():
        # print(df_.columns, row._asdict().keys())
        df_p, df_wc_p = preprocessing(pd.concat([df_, pd.DataFrame([row._asdict()])]), df_wc_)
        to_predict = df_p[features].tail(1)
        to_predict[categoricals] = to_predict[categoricals].astype("category")
        y = model.predict(to_predict)
        # print(round(y[0]))
        if target == "points":
            if y[0] == 3:
                games_.loc[row.Index, "home_score"] = 1
                games_.loc[row.Index, "away_score"] = 0
            elif y[0] == 1:
                games_.loc[row.Index, "home_score"] = 0
                games_.loc[row.Index, "away_score"] = 0
            else:
                games_.loc[row.Index, "home_score"] = 0
                games_.loc[row.Index, "away_score"] = 1
        else:
            if y[0] >= 0:
                games_.loc[row.Index, "home_score"] = round(y[0])
                games_.loc[row.Index, "away_score"] = 0
            else:
                games_.loc[row.Index, "home_score"] = 0
                games_.loc[row.Index, "away_score"] = -1*round(y[0])
    return games_
        
```
